## Random initialization for odd/even angles


```julia
function init_angles(max_layers::Int)
    odd = [2π .* rand(4) for _ in 1:max_layers]
    even = [2π .* rand(4) for _ in 1:max_layers]
    return odd, even
end
odd_angles, even_angles = init_angles(max_layers)
```
